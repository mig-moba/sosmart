# Compilar SOSmart a APK (Android) con Google Colab

Este notebook compila el proyecto SOSmart a un archivo `.apk` instalable en Android, sin necesidad de instalar nada en tu computadora.

**Pasos:**
1. Ejecuta las celdas de arriba hacia abajo, en orden (boton de play, o Shift+Enter).
2. En la celda 3 se te pedira subir `SOSmart_App.zip` (el proyecto comprimido).
3. La compilacion tarda entre 20 y 40 minutos la primera vez (descarga el Android SDK/NDK).
4. Al final se descarga automaticamente el `.apk` a tu computadora.
5. Copia el `.apk` a tu celular (por USB, Drive, WhatsApp, etc.) y abrelo para instalarlo. Puede que tengas que habilitar "Instalar apps de origenes desconocidos" en Android.

**Importante:** cada sesion de Colab es temporal. Si se desconecta o pasan varias horas, tendras que volver a ejecutar el notebook desde el inicio.

## 3. Subir el proyecto (SOSmart_App.zip)

Ejecuta esta celda y usa el boton "Elegir archivos" para subir `SOSmart_App.zip`.

```python
from google.colab import files

uploaded = files.upload()
zip_name = list(uploaded.keys())[0]
print(f"Subido: {zip_name}")
```

```python
!rm -rf SOSmart_App
!unzip -o -q "$zip_name" -d SOSmart_App
%cd SOSmart_App
!ls
```

## 4. Compilar el APK

Esta celda tarda mucho la primera vez (descarga Android SDK/NDK, ~20-40 min). Muestra mucho texto, es normal.

```python
!yes | buildozer -v android debug
```

## 5. Descargar el APK

```python
import glob
from google.colab import files

apk_files = glob.glob("bin/*.apk")
print("APK generado:", apk_files)

if apk_files:
    files.download(apk_files[0])
else:
    print("No se encontro ningun .apk. Revisa el log de la celda anterior para ver el error.")
```
